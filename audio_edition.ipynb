# 긴 시간의 WAV 파일 2초 단위로 분할

```python
def split_audio_into_chunks(file_path, chunk_length_ms=2000):
    
    # 1. 파일 불러오기
    print(f"파일을 불러오는 중...: {file_path}")
    audio = AudioSegment.from_file(file_path)

    # 2. 저장할 폴더 생성 (파일명과 동일한 이름)
    base_name = os.path.basename(file_path).split('.')[0]
    output_dir = f"./output_{base_name}"
    if not os.path.exists(output_dir):
        os.makedirs(output_dir)

    # 3. 자르기 루프
    duration_ms = len(audio)
    print(f"전체 길이: {duration_ms / 1000}초")

    count = 0
    for start_ms in range(0, duration_ms, chunk_length_ms):
        end_ms = start_ms + chunk_length_ms

        # 마지막 조각이 1초보다 짧을 경우 제외
        if end_ms > duration_ms:
            break

        chunk = audio[start_ms:end_ms]

        # 4. 파일 저장
        file_name = f"{base_name}_{count:04d}.wav"
        chunk.export(os.path.join(output_dir, file_name), format="wav")

        count += 1
        if count % 100 == 0:
            print(f"{count}개 파일 생성 완료...")

    print(f"✅ 작업 완료! 총 {count}개의 파일이 '{output_dir}' 폴더에 저장되었습니다.")
```

```python
split_audio_into_chunks(r"C:\Users\samsung\OneDrive\바탕 화면\Hornet\Hornet_1.wav")
```

# 10초 단위로 구성된 WAV 파일들을 2초 단위로 분할

```python
import os
from pydub import AudioSegment

def split_all_wavs_in_folder(input_folder, chunk_length_ms=2000):
    # 1. 폴더 내 파일 목록 가져오기
    files = [f for f in os.listdir(input_folder) if f.endswith(".wav")]

    print(f"총 {len(files)}개의 wav 파일 발견")

    for file in files:
        file_path = os.path.join(input_folder, file)
        print(f"\n처리 중: {file_path}")

        audio = AudioSegment.from_file(file_path)

        # 파일별 output 폴더 생성
        base_name = os.path.splitext(file)[0]
        output_dir = os.path.join(input_folder, f"output_{base_name}")
        os.makedirs(output_dir, exist_ok=True)

        duration_ms = len(audio)
        count = 0

        for start_ms in range(0, duration_ms, chunk_length_ms):
            end_ms = start_ms + chunk_length_ms

            if end_ms > duration_ms:
                break

            chunk = audio[start_ms:end_ms]

            file_name = f"{base_name}_{count:04d}.wav"
            chunk.export(os.path.join(output_dir, file_name), format="wav")

            count += 1

        print(f"{count}개 파일 생성 완료")

    print("\n✅ 모든 파일 처리 완료")
```

```python
split_all_wavs_in_folder(r"C:\Users\samsung\OneDrive\바탕 화면\No_Bee")
```

# 분할된 WAV 파일들을 다른 파일로 트랜스퍼

```python
import shutil

def collect_all_chunks(input_folder, archive_folder):
    # archive 폴더 생성
    os.makedirs(archive_folder, exist_ok=True)

    total_moved = 0

    # input_folder 안에 있는 것들 순회
    for item in os.listdir(input_folder):
        item_path = os.path.join(input_folder, item)

        # output_로 시작하는 폴더만 대상으로
        if os.path.isdir(item_path) and item.startswith("output_"):
            print(f"\n처리 중: {item_path}")

            for file in os.listdir(item_path):
                if file.endswith(".wav"):
                    src = os.path.join(item_path, file)

                    # 파일명 충돌 방지 (폴더 이름 붙이기)
                    new_name = f"{item}_{file}"
                    dst = os.path.join(archive_folder, new_name)

                    shutil.copy(src, dst)  # 복사 (원본 유지)
                    total_moved += 1

    print(f"\n✅ 완료! 총 {total_moved}개의 파일을 archive로 복사함")
```

```python
collect_all_chunks(
    r"C:\Users\samsung\OneDrive\바탕 화면\Hornet",
    r"C:\Users\samsung\OneDrive\바탕 화면\Hornet_data"
)
```

# 하이패스필터 적용

```python
import os
import numpy as np
import librosa
import soundfile as sf
from scipy.signal import butter, filtfilt

# ============================================
# 1. 폴더 설정
# ============================================
input_folder = r"C:\Users\samsung\OneDrive\바탕 화면\휴먼 프로젝트\audio\Bee"          # 입력 폴더
output_folder = r"C:\Users\samsung\OneDrive\바탕 화면\휴먼 프로젝트\audio2\Bee"        # 출력 폴더

# 출력 폴더 없으면 생성
os.makedirs(output_folder, exist_ok=True)

# ============================================
# 2. High-pass Filter 함수
# ============================================
def highpass_filter(data, cutoff, fs, order=5):
    nyquist = 0.5 * fs
    normal_cutoff = cutoff / nyquist
    
    b, a = butter(order, normal_cutoff, btype='high', analog=False)
    filtered_data = filtfilt(b, a, data)
    
    return filtered_data

# ============================================
# 3. 파라미터 설정
# ============================================
cutoff_freq = 150  # Hz

# ============================================
# 4. 폴더 내 모든 WAV 파일 처리
# ============================================
for file_name in os.listdir(input_folder):
    if file_name.endswith(".wav"):
        input_path = os.path.join(input_folder, file_name)
        output_path = os.path.join(output_folder, file_name)

        print(f"처리 중: {file_name}")

        # 오디오 로드
        audio, sr = librosa.load(input_path, sr=None)

        # 필터 적용
        filtered_audio = highpass_filter(audio, cutoff_freq, sr)

        # 저장
        sf.write(output_path, filtered_audio, sr)

print("모든 파일 High-pass filtering 완료!")
```
